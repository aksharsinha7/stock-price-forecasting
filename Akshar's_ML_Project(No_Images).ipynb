```python
all_brands = ['Nike', 'Adidas', 'Reebok', 'Converse', 'Puma', 'Vans', 'New Balance', 'Asics',
 'Fila', 'Skechers']

for x in all_brands:
  data[x] = np.nan
for i in range (len(data)):
  for y in all_brands:
    data_brands = list(data['Brand'])[i]
    if y in data_brands:
      data.loc[i,y] = 1
    else:
      data.loc[i,y] = 0
```

```python
print(data['Type'].unique())
```

```python
all_types= ['Basketball', 'Running', 'Casual', 'Lifestyle', 'Skate', 'Fashion', 'Walking',
 'Cross-training', 'Slides', 'Crossfit', 'Training', 'Trail',
 'Weightlifting', 'Hiking', 'Racing', 'Retro']
for x in all_types:
  data[x] = np.nan
for i in range (len(data)):
  for y in all_types:
    data_type = list(data['Type'])[i]
    if y in data_type:
      data.loc[i,y] = 1
    else:
      data.loc[i,y]= 0
```

```python
print(data['Color'].unique())
```

```python
all_colors = ['Black', 'Brown', 'White', 'Navy', 'Pink', 'Grey', 'Blue', 'Green', 'Yellow', 'Purple', 'Red', 'Checkered', 'Beige', 'Orange', 'Ivory', 'Egret','Gold', 'Multi-color', 'Burgundy', 'Zebra', 'Khaki', 'Charcoal', 'Cinder', 'Natural', 'Silver' , 'Gum']
for color in all_colors:
  data[color] = np.nan
for i in range (len(data)):
  for color in all_colors:
    data_color = list(data['Color'])[i]
    if color.lower() in data_color.lower():
      data.loc[i,color] = 1
    else:
      data.loc[i,color] = 0


```

```python
print(data['Material'].unique())
```

```python
 all_materials = ['Leather', 'Primeknit', 'Canvas', 'Mesh', 'Suede', 'Synthetic', "Textile", 'Knit', 'Nylon', 'Flexweave', 'Cushoning']
for mat in all_materials:
  data[mat] = np.nan
for i in range (len(data)):
  for mat in all_materials:
    data_materials = list(data['Material'])[i]
    if mat.lower() in data_materials.lower():
      data.loc[i,mat] = 1
    else:
      data.loc[i,mat] = 0
```

```python
data['Price (USD)'] = data['Price (USD)'].str.replace('$', '', regex=True).str.strip().astype(float)
data.head()
```

```python
#data = data.drop(['Model'], axis=1)
data = data.drop(['Color'], axis= 1)
#data = data.drop(['color'], axis = 1)
data = data.drop(['Type'], axis = 1)
data = data.drop(['Gender'], axis = 1)
data = data.drop(['Size'], axis = 1)
data = data.drop(['Material'], axis = 1)
```
